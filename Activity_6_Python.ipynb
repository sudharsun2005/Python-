```python
#While Loop : A while loop is used to repeatedly execute a block of code as long as a condition is true.
```

```python
i=1
while True:
  print("subiksha")
  i=i+1
  if i>5:
    break
```

```python
total_seats = 5

while total_seats > 0:
    print("Seat Booked Successfully")
    print("Remaining Seats:", total_seats - 1)
    total_seats -= 1

print("All Seats Are Filled")
```

```python
total_chocolates = 5

while total_chocolates > 0:
    print("Chocolate Eaten Successfully!")
    print("Remaining Chocolates:", total_chocolates - 1)
    total_chocolates -= 1

print("All Chocolates Are Finished!")
```

```python
# For loop
# For Loop : A for loop is used to execute a block of code repeatedly for a fixed number of times.
```

```python
print("Available Seat Numbers:")

for seat in range(1, 11):
    print("Seat Number:", seat)
```

```python
sum = 0
for i in range(1, 6):
  sum=sum+i
print(sum)
```

```python
n = int(input("Enter a number: "))
sum = 0
for i in range(1, n+1):
  sum=sum+i
print(sum)
```
