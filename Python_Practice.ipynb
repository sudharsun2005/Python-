```python
# Fruit prices per kg
apple_price = 150
banana_price = 50
orange_price = 80

# Fruit weights in kg
apple_weight = 2.5
banana_weight = 1.5
orange_weight = 3.0

# Calculate totals
apple_total = apple_price * apple_weight
banana_total = banana_price * banana_weight
orange_total = orange_price * orange_weight

total = apple_total + banana_total + orange_total

print("-----------FRUIT BILL---------------")

print("Apples  :\t\t", apple_total)
print("Bananas :\t\t", banana_total)
print("Oranges :\t\t", orange_total)
print("Total   :\t\t", total)

print("------------------------------------")
```
